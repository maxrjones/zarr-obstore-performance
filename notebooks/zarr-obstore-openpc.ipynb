```python
import zarr
from obstore.auth.planetary_computer import PlanetaryComputerCredentialProvider
from obstore.store import AzureStore
from zarr.storage import ObjectStore
```

```python
zarr.config.set({"async.concurrency": 128})
```

```python
url = "https://azureopendatastorage.blob.core.windows.net/noaa/conus404.zarr"
```

```python
azstore = AzureStore(
    credential_provider=PlanetaryComputerCredentialProvider(url),
    client_options={"timeout": "10m", "connect_timeout": "10m"},
)
store = ObjectStore(azstore, read_only=True)
```

```python
arr = zarr.open_array(store, path="T2", mode="r")
```

```python
%%time
subset = arr[0, 0, 0]
```
